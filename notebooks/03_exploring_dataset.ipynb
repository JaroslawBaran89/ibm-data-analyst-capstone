Print the mean age of the survey participants.


First let's map age ranges to approximate numeric values


```python
age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

# Create a new numeric Age column
df['Age_numeric'] = df['Age'].map(age_mapping)
```

## Technology trends summary

After removing duplicates, split the multi-answer columns on `;` and count the top 10 technologies used now and wanted next year, plus demographic breakdowns. These figures feed the charts, the dashboard and the final presentation.


```python
import pandas as pd, json
df = pd.read_csv(file_path).drop_duplicates()

def top(col, n=10):
    s = df[col].dropna().str.split(';').explode().str.strip()
    return s.value_counts().head(n).to_dict()

out = {"rows": len(df)}
for c in ["LanguageHaveWorkedWith","LanguageWantToWorkWith",
          "DatabaseHaveWorkedWith","DatabaseWantToWorkWith",
          "PlatformHaveWorkedWith","PlatformWantToWorkWith",
          "WebframeHaveWorkedWith","WebframeWantToWorkWith"]:
    out[c] = top(c)
out["Age"] = df["Age"].value_counts().to_dict()
out["EdLevel"] = df["EdLevel"].value_counts().to_dict()
out["Country"] = df["Country"].value_counts().head(15).to_dict()
out["RemoteWork"] = df["RemoteWork"].value_counts().to_dict()
out["Comp_median"] = df["ConvertedCompYearly"].median()
print(json.dumps(out, ensure_ascii=False))
```
